## 残高取得

### RPCノードの設定
制限が緩めで安定しているパブリックRPCノードを使用

```python
RPC_URL = "https://eth.drpc.org"
print(f"RPC URL: {RPC_URL}")
```

### User-Agentヘッダーの設定
パブリックノードのBot弾き（WAF）を回避するため、ブラウザのUser-Agentを設定する

```python
headers = {
    'User-Agent': 'Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/120.0.0.0 Safari/537.36'
}
print("User-Agent ヘーダー設定完了")
```

### Web3接続のテスト
プロバイダの初期化（ヘッダーとタイムアウトを設定）

```python
print(f" {RPC_URL} へ接続します...\n")

try:
    w3 = Web3(Web3.HTTPProvider(RPC_URL, request_kwargs={'headers': headers, 'timeout': 5}))
    
    # 最新のブロック情報を取得
    latest_block = w3.eth.block_number
    print(f" 接続成功！ (現在の最新ブロック: {latest_block})\n")
    
except Exception as e:
    print(f" 接続失敗: {type(e).__name__} - {e}")
```

### アドレスの残高を取得


```python
# アドレスの残高を取得
print("--- アドレス残高の確認 ---")

try:
    # イーサリアム共同創設者 Vitalik Buterin 氏の公開アドレス
    target_address = "0xd8dA6BF26964aF9D7eEd9e03E53415D37aA96045"
    
    # get_balance で残高を取得（単位は最小単位の「Wei」）
    balance_wei = w3.eth.get_balance(target_address)
    
    # from_wei で Wei を「ETH」表記に変換
    balance_eth = w3.from_wei(balance_wei, "ether")
    
    print(f"Target : {target_address}")
    print(f"Balance: {balance_eth:.4f} ETH")
    
except Exception as e:
    print(f"[error] データ取得中にエラー発生: {e}")
```
