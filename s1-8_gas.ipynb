## 実装のポイント：`Gwei` という単位

ETHの最小単位は「Wei」ですが、Gas代を表すときはゼロが多すぎて読みにくいため、「1 Gwei = $10^9$ Wei」という中間単位を使うのがWeb3エンジニアの常識です。

```python
from web3 import Web3

# 1. 接続設定
RPC_URL = "https://eth.drpc.org"
headers = {'User-Agent': 'Mozilla/5.0 (Windows NT 10.0; Win64; x64)'}
w3 = Web3(Web3.HTTPProvider(RPC_URL, request_kwargs={'headers': headers, 'timeout': 5}))

print(" 現在のGas代（ネットワーク手数料）をチェックします...\n")

try:
    # 2. 現在のGas価格を取得（単位は Wei）
    gas_price_wei = w3.eth.gas_price
    
    # 3. Wei を Gwei に変換
    gas_price_gwei = w3.from_wei(gas_price_wei, 'gwei')
    
    print(" 取得成功！")
    print("-" * 40)
    print(f"現在のGas代: {gas_price_gwei:.2f} Gwei")
    print(f"(Wei表記: {gas_price_wei} Wei)")
    print("-" * 40)
    
except Exception as e:
    print(f" エラーが発生しました: {e}")
```
