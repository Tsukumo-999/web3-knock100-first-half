## §1-5：ENS（イーサリアム・ネーム・サービス）を名前解決せよ

これまでのノートで扱ってきたウォレットアドレスは `0xd8dA6BF26964aF9D7eEd9e03E53415D37aA96045` のような、人間には到底覚えられない42文字の英数字の羅列でした。

この不便さを解消するためにWeb3の世界で広く使われているのが**ENS（Ethereum Name Service）**です。
インターネットのURL（DNS）と同じように、複雑なアドレスを `vitalik.eth` のような分かりやすい名前に紐付けることができます。

この問題では、Web3.pyに組み込まれたENS機能を使って、人間が読める名前から実際のアドレスを割り出す「名前解決」を体験します。

## ENSを体験する `.ens` モジュールの活用

ENSはスマートコントラクト上で動いていますが、世界中で頻繁に使われる標準規格であるため、Web3.pyには専用の便利なモジュール（`w3.ens`）が最初から用意されています。
複雑なABIを定義しなくても、わずか数行で名前解決が可能です。

### 1. 接続設定

```python
from web3 import Web3

# 1. 接続設定（ENSはイーサリアム・メインネット上で稼働しているため、メインネットのRPCが必要です）
RPC_URL = "https://eth.drpc.org"
headers = {'User-Agent': 'Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36'}
w3 = Web3(Web3.HTTPProvider(RPC_URL, request_kwargs={'headers': headers, 'timeout': 5}))
```

### 2. ENSでaddress取得
ENSに登録されているドメインからaddressを引いてみます。<br>

web2(webページなど)のaddressと比較<br>
web2(dns) : google.com  → 142.250.23.94<br>
web3(ens) : vitalik.eth → 0xd8dA6BF26964aF9D7eEd9e03E53415D37aA96045<br>

```python

print("ENS名前解決のサンプルコード")
print("-" * 40)

# 2. 名前解決したいENSドメイン（.eth）
# 例として、イーサリアムの共同創設者 Vitalik Buterin 氏のドメインを使用
target_domain = "vitalik.eth"

try:
    # 3. ENSから実際のアドレスを取得 (Forward Resolution)
    # これだけで、裏側でコントラクトへの問い合わせを自動で行ってくれます
    resolved_address = w3.ens.address(target_domain)
    
    if resolved_address:
        print(f" 解決成功: {target_domain}")
        print(f" 実際のアドレス: {resolved_address}")
        
        # せっかくなので、そのアドレスの残高も取得してみましょう
        bal_wei = w3.eth.get_balance(resolved_address)
        bal_eth = w3.from_wei(bal_wei, "ether")
        print(f" ウォレット残高: {bal_eth:,.2f} ETH")
    else:
        print(f" {target_domain} はどのアドレスにも紐付いていないか、未登録です。")

except Exception as e:
    print(f" 名前解決中にエラーが発生しました: {e}")
    
print("-" * 40)
```

### コラム：逆引き（リバースレゾリューション）について
アドレスからENSドメインを割り出すことを「逆引き」と呼びます。
Web3.pyでは `w3.ens.name('0xd8dA...')` と記述するだけで、そのアドレスの所有者が設定しているメインの `.eth` ドメインを取得することも可能です。（※所有者がリバースレコードを設定している場合のみ有効です）
